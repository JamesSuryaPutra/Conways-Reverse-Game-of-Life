```python
import numpy as np
import pandas as pd

from tensorflow.keras.layers import Dense, Dropout, Activation, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras.models import Sequential
from tensorflow.keras.utils import normalize

import os
print(os.listdir("../input"))
```

# Analyze data

```python
train = pd.read_csv("/kaggle/input/competitions/conway-s-reverse-game-of-life/train.csv.zip")
test = pd.read_csv("/kaggle/input/competitions/conway-s-reverse-game-of-life/test.csv.zip")
```

```python
X = train[[col for col in train.columns if "stop" in col]]
y = train[[col for col in train.columns if "start" in col]]
```

```python
train.head()
```

```python
test.head()
```

# Normalize and reshape arrays

```python
X = np.array(X)
y = np.array(y)
```

```python
X = normalize(X)
y = normalize(y)
```

```python
X = X.reshape(-1, 20, 20, 1)
```

# Model training

```python
model = Sequential()

# First layer
model.add(Conv2D(64, (3, 3), activation='relu', input_shape=X.shape[1:]))
model.add(MaxPooling2D(pool_size=(2, 2)))
# Second layer
model.add(Conv2D(64, (3, 3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Flatten())
model.add(Dense(64, activation='relu'))

# Softmax layer
model.add(Dense(400, activation='softmax'))

# Compile the model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# Fit the model
model.fit(
    X,y,
    epochs=10,
    batch_size=32,
    validation_split=0.1
)
```

# Submission

```python
test.head()
```

```python
x_test = test.drop(['id', 'delta'], axis=1)
```

```python
x_test = np.array(x_test).reshape(-1, 20, 20, 1)

normalize(x_test)
```

```python
predictions = model.predict(x_test)
```

```python
predictions = predictions.round()
predictions = predictions.astype(int)

column_names = ['start.' + str(i) for i in range(1, 401)]
```
